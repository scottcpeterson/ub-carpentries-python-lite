## Things to know about JupyterLite

- **No shell commands.** Lines starting with `!` (such as `!pip install` or `!ls`) don't work here. Everything the lessons need is preinstalled: `pandas`, `numpy`, `matplotlib`, `seaborn`, `plotly`.
- **Your work lives in your browser.** Notebooks you create are saved to this browser's local storage, not a server. Clearing site data wipes them. Right-click a file and choose Download to keep anything.
- **First load is slow.** The Python kernel downloads on first visit, then it's cached.
