- Control a program with lists, loops, conditions, and your own functions.

## How to use JupyterLite

- **File browser (left).** Open the `data/` folder to see the lesson files. The paths in the lesson, like `data/2011_circ.csv`, work as written. You do not need to download or unzip anything.
- **Run a cell.** Click a code cell, then press **Shift+Enter**. The output appears below the cell.
- **Start a new notebook.** Click **+** at the top of the file browser, then choose **Python (XPython)** under Notebook.
- **If something hangs.** Use the **Kernel** menu, then **Restart Kernel**. Then run your cells again from the top.
- **Save and keep your work.** Notebooks are saved in this browser's local storage, not on a server. Clearing site data deletes them. Right-click a file in the file browser and choose **Download** to keep a copy.
- **Add your own files.** Click the upload arrow at the top of the file browser.
- **No shell commands.** Lines that start with `!`, such as `!pip install` or `!ls`, do not work. The packages the lesson needs are already installed: `pandas`, `numpy`, `matplotlib`, `seaborn`, `plotly`.
- **First load is slow.** The Python kernel downloads on your first visit. After that it is cached.
